```python
import pandas as pd
import joblib
import os
from collections import Counter

print("--- Starting Final Offline ML Skill Extraction Process ---")

# 1. Load the dataset
try:
    df = pd.read_csv('../Datasets/Skill_Recommendation/job_dataset.csv')
    print("'job_dataset.csv' loaded successfully!")
except FileNotFoundError:
    print("Error: 'job_dataset.csv' not found. Please ensure it's in the correct folder.")
    exit()

# 2. Clean the data
df.dropna(subset=['Title', 'Skills'], inplace=True)
print(f"Total jobs to analyze: {len(df)}")

# 3. **THE FIX:** Automatically identify the top 25 most frequent job titles from the data
top_job_titles = df['Title'].value_counts().nlargest(25).index.tolist()
print("\nFound the top 25 most common job titles in the dataset:")
print(top_job_titles)


# 4. Run the ML "Extraction" process for these actual job titles
ml_generated_skills = {}
for job_title in top_job_titles:
    print(f"Processing skills for: {job_title}...")
    
    # Filter for the exact job title
    filtered_df = df[df['Title'] == job_title]
    
    # Get all skills from the 'Skills' column for this job title
    all_skills_for_this_job = []
    for skills_string in filtered_df['Skills']:
        # The skills are separated by semicolons in this dataset
        all_skills_for_this_job.extend([skill.strip() for skill in skills_string.split(';')])
    
    # Count the frequency of each skill to find the most common ones
    skill_counts = Counter(all_skills_for_this_job)
    top_skills = [skill for skill, count in skill_counts.most_common(10)]
    
    ml_generated_skills[job_title] = top_skills

# 5. Save the final, ML-generated dictionary
output_dir = '../Models/Skill_Recommendation'
os.makedirs(output_dir, exist_ok=True)
output_path = os.path.join(output_dir, 'ml_generated_skills.joblib')
joblib.dump(ml_generated_skills, output_path)

print(f"\nSuccessfully processed all jobs. ML-generated skills saved to: {output_path}")
print("\n--- Final Generated Skills Dictionary ---")
print(ml_generated_skills)

```
