```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import joblib
import os
import warnings

warnings.filterwarnings('ignore')

print("--- Starting ML Course Recommender Model Build ---")

# 1. Load the dataset
try:
    df = pd.read_csv('../Datasets/Course_Recommendation/MOOC.csv', encoding='latin-1')
    print("'MOOC.csv' dataset loaded successfully!")
except FileNotFoundError:
    print("Error: 'MOOC.csv' not found. Please ensure it's in the correct directory.")
    exit()

# 2. Data Cleaning & Feature Engineering
print("\n--- Cleaning and Preparing Data ---")
df.dropna(subset=['Course Name', 'Course Description', 'all_skill'], inplace=True)
df.drop_duplicates(subset='Course Name', inplace=True)
df.reset_index(drop=True, inplace=True)

# Create a combined 'tags' column for TF-IDF vectorization
df['tags'] = df['Course Name'] + ' ' + df['Course Description'] + ' ' + df['all_skill']
print("Combined 'tags' feature created.")

# 3. ML Model Training (TF-IDF + Cosine Similarity)
print("\n--- Building TF-IDF Vectorizer ---")
# This learns the vocabulary and importance of words from the course content
tfidf = TfidfVectorizer(max_features=5000, stop_words='english')
vectors = tfidf.fit_transform(df['tags']).toarray()
print(f"Text vectorized. Shape: {vectors.shape}")

print("\n--- Calculating Cosine Similarity Matrix (The Model) ---")
# This matrix stores the similarity score between every pair of courses
similarity_matrix = cosine_similarity(vectors)
print(f"ML Similarity Matrix built. Shape: {similarity_matrix.shape}")

# 4. Save the processed data and the ML model (similarity matrix)
output_dir = '../Models/Course_Recommendation'
os.makedirs(output_dir, exist_ok=True)

# These are the files your app.py will load
df_path = os.path.join(output_dir, 'courses_processed.joblib')
matrix_path = os.path.join(output_dir, 'similarity_matrix.joblib')
vectorizer_path = os.path.join(output_dir, 'tfidf_vectorizer.joblib') # Also save the vectorizer

joblib.dump(df[['Course Name']], df_path) # Save just the course names
joblib.dump(similarity_matrix, matrix_path) # Save the ML similarity model
joblib.dump(tfidf, vectorizer_path) # Save the TF-IDF model

print(f"\nProcessed data and ML models saved successfully in: {output_dir}")
```
